#### 7. Remove all duplicate characters from a string.

```python
def remove_dup(mystr):
    """It will remove all the duplicates from the given string"""
    result = ""
    for i in mystr:
        if i not in result:
            result+=i
    return result
```

```python
str1 = "pyhton python"
print("original string:",str1)
res = remove_dup(str1)
print("after removing duplicates is:",res)
```

### 11. Flatten a nested list.

```python
def flatten_list(nested):
    flat_list = []
    for i in nested:
        if isinstance(i, list):
            flat_list += flatten_list(i)
        else:
            flat_list.append(i)
    return flat_list
```

```python
nested = [10,20,[200,300],400]
print(flatten_list(nested))
```

```python
nested = [10, 20, [200, 300], 400]
flat = []
for i in nested:
    if type(i) == int:
        flat.append(i)
    else:
        flat.extend(i)
print(flat)
```

#### 30. Print all prime numbers between 1 to 100 using for loop.

```python
primes = []  
for num in range(2, 101):
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        primes.append(num)  
print(primes)  
```

#### 35. Write a function to check if a number is an Armstrong number.

```python
def armstrong(n):
    num = str(n)
    pwr = len(num)
    total = 0
    for i in num:
        total += int(i) ** pwr
    return total == n

n = int(input("Enter the number: "))
if armstrong(n):
    print("Yes, it's an Armstrong number")
else:
    print("No, it's not an Armstrong number")
```

### 38. Write a function to find GCD and LCM of two numbers.

```python
def gcd_lcm(a,b):
    def gcd(x,y):
        while y:
            x,y=y, x%y
        return x
    gcd_val = gcd(a,b)
    lcm_val = (a*b)//gcd_val

    return gcd_val,lcm_val

input1 = int(input("enter a first num:"))
input2 = int(input("enter a second number"))

gcd_result,lcm_result = gcd_lcm(input1,input2)
print("GCD:",gcd_result)
print("LCM:",lcm_result)
```

### 50. Demonstrate exception handling with file operations.

```python
def file_operations():
    try:
        file_name = input("Enter the filename to read: ")
        with open(file_name, 'r') as file:
            content = file.read()
            print("File content:\n", content)
    except FileNotFoundError:
        print(f"Error: The file '{file_name}' does not exist.")
    except PermissionError:
        print(f"Error: You do not have permission to read the file '{file_name}'.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    finally:
        print("File read operation completed.\n")

    try:
        with open("output.txt", "w") as file:
            file.write("This is a sample output written to the file.\n")
            print("Data written to 'output.txt' successfully.")

    except IOError:
        print("Error occurred while writing to the file.")

    finally:
        print("File write operation completed.")

file_operations()
```
